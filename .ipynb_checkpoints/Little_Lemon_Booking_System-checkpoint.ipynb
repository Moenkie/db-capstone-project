```python
import mysql.connector as connector
```

```python
#create a connection to the little lemon database
connection = connector.connect(user = 'User1', password = '12345678', db = 'littlelemondb')
```

```python
#create a cursor
cursor = connection.cursor()
```

```python
#run the show tables query via the connection created, fetch the results and print
show_tables_query = """SHOW tables"""
cursor.execute(show_tables_query)
results = cursor.fetchall()
print(results)
```

```python
#run the a query returning the customers whose order are greater than 60 dollars for a promotional campaign via the connection created, fetch the results and print
promotional_campaign = """
SELECT c.customer_name, c.contact_number
FROM customers as c
INNER JOIN orders as o
ON c.CustomerID = o.CustomerID
WHERE totalcost > 60;
"""
cursor.execute(promotional_campaign)
results = cursor.fetchall()
print(results)
```

```python
#close cursor and connection is still open
if connection.is_connected():
 cursor.close()
 print("The cursor is closed.")
 connection.close()
 print("Connection is closed.")
else:
 print("Connection is already closed")
```
